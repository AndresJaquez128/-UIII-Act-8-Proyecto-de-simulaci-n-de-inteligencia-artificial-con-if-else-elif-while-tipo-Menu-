###Genera un programa en Python que simule inteligencia artificial utilizando estructuras condicionales `if`, `if else`, `elif` y el ciclo `while`. El programa debe estar relacionado con un dibujante y lápices, usando un menú interactivo para el usuario.

##Requisitos:

* No utilizar programación orientada a objetos.
* No utilizar funciones.
* No utilizar ciclos `for`.
* No utilizar `try`.
* El programa debe ser interactivo y sencillo.
* Debe incluir varias opciones de menú.
* Debe usar decisiones condicionales para simular recomendaciones de inteligencia artificial.
* El código debe ser atractivo y fácil de entender para alumnos de preparatoria.
* Utilizar comentarios para separar secciones del programa.


```python
print("======================================================")
print("        IA PARA DIBUJANTE A LÁPIZ")
print("======================================================")

nombre = input("Ingrese el nombre del dibujante: ")

while True:

    print("\n=================== MENÚ ===================")
    print("1. Analizar tipo de dibujo")
    print("2. Analizar lápiz")
    print("3. Simular dibujo")
    print("4. Salir")
    print("============================================")

    opcion = input("Seleccione una opción: ")

    # ==================================================
    # OPCIÓN 1
    # ==================================================

    if opcion == "1":

        print("\n========== TIPO DE DIBUJO ==========")

        print("1. Retrato")
        print("2. Paisaje")
        print("3. Anime")

        dibujo = input("Seleccione una opción: ")

        if dibujo == "1":

            print("\nLa IA detectó dibujo de retrato.")
            print("Recomendaciones:")
            print("- Dibujar ojos primero")
            print("- Usar sombras suaves")
            print("- Mantener proporciones")

        elif dibujo == "2":

            print("\nLa IA detectó dibujo de paisaje.")
            print("Recomendaciones:")
            print("- Dibujar horizonte")
            print("- Agregar profundidad")
            print("- Usar sombras ligeras")

        elif dibujo == "3":

            print("\nLa IA detectó dibujo estilo anime.")
            print("Recomendaciones:")
            print("- Ojos grandes")
            print("- Líneas marcadas")
            print("- Expresiones exageradas")

        else:

            print("\nOpción incorrecta")

    # ==================================================
    # OPCIÓN 2
    # ==================================================

    elif opcion == "2":

        print("\n========== ANÁLISIS DEL LÁPIZ ==========")

        print("1. Lápiz HB")
        print("2. Lápiz 2B")
        print("3. Lápiz 6B")

        lapiz = input("Seleccione una opción: ")

        if lapiz == "1":

            print("\nLa IA detectó lápiz HB.")
            print("Ideal para:")
            print("- Bocetos")
            print("- Líneas suaves")
            print("- Dibujos escolares")

        elif lapiz == "2":

            print("\nLa IA detectó lápiz 2B.")
            print("Ideal para:")
            print("- Sombras")
            print("- Dibujos detallados")
            print("- Retratos")

        elif lapiz == "3":

            print("\nLa IA detectó lápiz 6B.")
            print("Ideal para:")
            print("- Sombras oscuras")
            print("- Efectos intensos")
            print("- Contraste fuerte")

        else:

            print("\nOpción incorrecta")

    # ==================================================
    # OPCIÓN 3
    # ==================================================

    elif opcion == "3":

        print("\n========== SIMULADOR DE DIBUJO ==========")

        tiempo = int(input("Ingrese horas de dibujo: "))
        detalle = input("¿El dibujo tiene mucho detalle? (si/no): ")

        if tiempo >= 5 and detalle == "si":

            print("\nLa IA detectó un dibujo avanzado.")
            print("Recomendaciones:")
            print("- Descansar la mano")
            print("- Mantener precisión")
            print("- Usar sombras suaves")

        elif tiempo >= 2 and detalle == "no":

            print("\nLa IA detectó dibujo básico.")
            print("Recomendaciones:")
            print("- Practicar líneas")
            print("- Mejorar proporciones")

        else:

            print("\nLa IA detectó dibujo intermedio.")
            print("Recomendaciones:")
            print("- Mejorar detalles")
            print("- Practicar sombras")

    # ==================================================
    # SALIR
    # ==================================================

    elif opcion == "4":

        print("\nGracias por usar la IA para Dibujantes,", nombre)
        print("Programa finalizado.")
        break

    # ==================================================
    # ERROR
    # ==================================================

    else:

        print("\nOpción incorrecta")
print("Programa hecho por Jaquez Andres 0074")
```
